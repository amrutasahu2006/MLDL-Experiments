```python
# Cell 1: Install & import everything
!pip install tensorflow matplotlib seaborn --quiet

import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import classification_report, confusion_matrix
import time

print("TensorFlow version:", tf.__version__)
```

```python
# Cell 2: Load Fashion-MNIST
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()

print("Dataset loaded successfully!")
print("Train shape:", x_train.shape, "Labels:", y_train.shape)
print("Test shape:", x_test.shape, "Labels:", y_test.shape)
```

```python
# Cell 3: Preprocess
x_train = x_train / 255.0
x_test  = x_test  / 255.0

# Add channel dimension for CNN (28,28) → (28,28,1)
x_train = np.expand_dims(x_train, axis=-1)
x_test  = np.expand_dims(x_test,  axis=-1)

print("Preprocessing done. New shape:", x_train.shape)
```

```python
# Cell 4: Build CNN
model = models.Sequential([
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(28,28,1)),
    layers.MaxPooling2D((2,2)),
    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D((2,2)),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(10, activation='softmax')
])

model.summary()
```

```python
# Cell 5: Compile and train
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

start_time = time.time()
history = model.fit(x_train, y_train,
                    epochs=20,
                    batch_size=128,
                    validation_split=0.1,
                    verbose=1)
training_time = time.time() - start_time
print(f"Training completed in {training_time:.2f} seconds")
```

```python
# Cell 6: Evaluation
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=0)
print(f"Test Accuracy: {test_acc*100:.2f}%")

# Predictions
y_pred = np.argmax(model.predict(x_test), axis=1)

# Classification report
print(classification_report(y_test, y_pred, target_names=[
    "T-shirt/top","Trouser","Pullover","Dress","Coat",
    "Sandal","Shirt","Sneaker","Bag","Ankle boot"]))
```

## 6. Performance Analysis

### Interpretation of Results

From the evaluation metrics and visualizations, we can conclude the following:

*   **Overall Accuracy:** The model achieved a test accuracy of **91.26%**, which is quite good for the Fashion-MNIST dataset.
*   **Training History:**
    *   The **accuracy plots** show that both training and validation accuracy increased steadily, indicating that the model was learning effectively. The validation accuracy generally tracked closely with the training accuracy, suggesting good generalization.
    *   The **loss plots** show a similar trend, with both training and validation loss decreasing over epochs. Towards the end, there might be a slight divergence or flattening, which could indicate the model is starting to converge or perhaps lightly overfitting, though the validation loss remains low.
*   **Classification Report:**
    *   **High Precision/Recall/F1-score** are observed for classes like 'Trouser', 'Sandal', 'Sneaker', 'Bag', and 'Ankle boot'. This means the model is very good at identifying these items and rarely confuses them with others.
    *   **Lower Performance for 'Shirt' and 'Pullover'**: The 'Shirt' class has the lowest F1-score (0.75), indicating that it is often misclassified. 'Pullover' and 'Coat' also have slightly lower scores compared to other classes.
*   **Confusion Matrix Insights:**
    *   The confusion matrix visually confirms the strengths and weaknesses. It highlights that 'Shirt' is frequently confused with 'T-shirt/top', 'Pullover', and 'Coat', which are visually similar clothing items. Similarly, 'Pullover' and 'Coat' might be confused with each other.
    *   There are very few misclassifications between distinct categories like 'Sandal' and 'Trouser', as expected.

In summary, the CNN model performs well on the Fashion-MNIST dataset, demonstrating strong generalization capabilities. The main areas for potential improvement lie in distinguishing between visually similar clothing items, particularly 'Shirt' from 'T-shirt/top', 'Pullover', and 'Coat'.

## 7. Hyperparameter Tuning

Hyperparameter tuning is the process of finding the optimal set of hyperparameters for a machine learning model to achieve the best performance. Instead of manually guessing, we can use automated methods to search through a defined range of hyperparameter values.

For this CNN model, we can tune hyperparameters such as the number of filters in convolutional layers, kernel sizes, number of dense units, dropout rate, and learning rate. We will use `keras_tuner` for this task.

### Impact of Hyperparameter Tuning

After running the hyperparameter tuning process, we observe the following impacts:

*   **Improved Performance (Potentially):** The `best_model` derived from tuning typically achieves a validation accuracy that is equal to or better than the baseline model, as the tuner explores various combinations to find an optimal set. The reported `Best Model Test Accuracy` indicates the performance of the tuned model.
*   **Identification of Optimal Configuration:** The tuning process provides insights into which hyperparameter values work best for this specific dataset and model architecture. For example, it suggests optimal filter sizes, dense layer units, dropout rates, and learning rates.
*   **Reduced Overfitting/Underfitting:** By adjusting parameters like dropout rate and learning rate, tuning can help in achieving a better balance between bias and variance, potentially leading to a model that generalizes better to unseen data and avoids both severe overfitting and underfitting.
*   **Computational Cost:** Hyperparameter tuning is computationally intensive, as it involves training multiple models. The `max_trials` and `executions_per_trial` parameters directly influence the time and resources required.

In summary, hyperparameter tuning is a crucial step to maximize the model's performance by systematically exploring the hyperparameter space. The optimal hyperparameters found provide a more robust and better-performing model compared to one built with arbitrary parameter choices.
